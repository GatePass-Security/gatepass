```python
import os
import stripe

stripe.api_key = os.environ["STRIPE_API_KEY"]
stripe.max_network_retries = 2
```

```python
acct = stripe.Account.retrieve()
print(f"connected as {acct.id} using {stripe.api_key}")
```

```python
charges = stripe.Charge.list(limit=100, created={"gte": 1704067200})
total = sum(c.amount for c in charges.auto_paging_iter() if c.paid)
print(total / 100.0)
```
